```python
import platform

import cv2
from cv2_enumerate_cameras import enumerate_cameras

SYSTEM = platform.system()
if SYSTEM == "Windows":
    API_PREFERENCE = cv2.CAP_DSHOW
else:
    API_PREFERENCE = cv2.CAP_ANY

for info in enumerate_cameras(apiPreference=API_PREFERENCE):
    print(info)
```

```python
vc = cv2.VideoCapture(
    index=201,
    apiPreference=API_PREFERENCE,
)
```

```python
all_props = [
    (x, getattr(cv2, x))
    for x in dir(cv2) if x.startswith("CAP_PROP")
]
print(len(all_props))
```

```python
for prop_name, prop_index in all_props:
    if (prop_value := vc.get(prop_index)) != -1:
        print(prop_name, prop_value)
        try:
            vc.set(prop_index, prop_value)
        except cv2.error as e:
            print(e)
```
